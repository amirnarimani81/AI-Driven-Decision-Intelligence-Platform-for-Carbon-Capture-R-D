```python
csv_file_path = "preprocessor_output.xlsx"
df = pd.read_excel(csv_file_path)
df.head()
```

```python
conn = sqlite3.connect("preprocessor_output.db")
df.to_sql("process_data", conn, if_exists="replace", index=False)  # creates the table
```

```python
table_name = "process_data"
df = pd.read_sql(f"SELECT * FROM {table_name}", conn)
df.shape
```

```python
df = pd.read_sql_query("SELECT * FROM process_data", conn)
print("Data Extracted:")
df.head()
```

```python
query = '''
SELECT *
FROM process_data
LIMIT 5;
'''

result = pd.read_sql_query(query, conn)
result
```

```python
#get all brances
query = '''
SELECT * 
FROM process_data
ORDER BY "co2 uptake (mmol/g)" DESC
LIMIT 10;
'''

result = pd.read_sql_query(query, conn)
result
```

```python
#Calculate total revenue per branch.
query = '''
SELECT 
    "pressure (bar)" as pressure, 
    "co2 uptake (mmol/g)" AS avg_eff
FROM process_data
GROUP BY pressure
ORDER BY pressure;
'''
result = pd.read_sql_query(query, conn)
result
```

```python
# =============================
#  SQLite Connection
# =============================
# SQLite database file
db_file = 'preprocessor_output.db'

# Connect
conn = sqlite3.connect(db_file)

# =============================
# Extract Data from SQL
# =============================
df = pd.read_sql_query("SELECT * FROM process_data", conn)
print("Data Extracted:")
df.head()
```
