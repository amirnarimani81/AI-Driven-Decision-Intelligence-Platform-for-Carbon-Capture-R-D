  <p style="text-align: justify; font-family: cursive; font-size: 14px; color: green; font-weight: bold">
    Introduction 
  </p>

  <p>
    Predicting gas uptake in Metal–Organic Frameworks (MOFs) is a critical challenge in gas storage and separation applications. Experimental adsorption measurements under varying temperature and pressure are costly and time-consuming, which limits rapid screening of new MOF materials. This project develops a machine learning framework to predict CO₂ uptake (wt%) in MOFs using experimental data. 
    
  </p>

  <p style="text-align: justify; font-family: cursive; font-size: 14px; color: green; font-weight: bold">
    🎯 Objective
  </p>

  <p>
    The objective of this project is to develop and benchmark machine learning models for predicting CO₂ uptake (wt%) in MOFs. The models should:
  </p>

  <ul>
    <li>work effectively with limited and heterogeneous data,</li>
    <li>prevent data leakage,</li>
    <li>and generalize to unseen MOFs.</li>
  </ul>

  <p style="text-align: justify; font-family: cursive; font-size: 14px; color: green; font-weight: bold">
    🔍 Issues and Challenges
  </p>

  <ul>
    <li><b>Multiple samples per MOF:</b> Each MOF has multiple measurements under different temperature and pressure conditions, requiring GroupKFold splitting.</li>
    <li><b>Skewed target distribution:</b> Gas uptake is skewed; log transformation is needed for stable regression.</li>
    <li><b>Multicollinearity:</b> Some descriptors are highly correlated, which may lead to overfitting.</li>
    <li><b>Limited dataset size:</b> Only 589 samples, so robust validation and feature selection are essential.</li>
  </ul>

  <p style="text-align: justify; font-family: cursive; font-size: 14px; color: green; font-weight: bold">
    🛠️ Proposed Solution
  </p>

  <p>
    A complete two method without pipeline and using pipeline/ColumnTransformer will be designed with the following steps:
  </p>

  <p style="text-align: justify; font-family: cursive; font-size: 14px; color: green; font-weight: bold">
    1) Data Processing
  </p>

  <ul>
    <li>Remove duplicates and missing values</li>
    <li>Apply log transformation to the target to reduce skewness</li>
    <li>Scale features for models sensitive to feature scale (SVR, MLP)</li>
  </ul>

  <p style="text-align: justify; font-family: cursive; font-size: 14px; color: green; font-weight: bold">
    2) Feature Engineering
  </p>

  <ul>
    <li>Create physics-informed features: <b>log(P)</b>, <b>1/T</b>, <b>P × T</b></li>
    <li>Generate interaction terms to capture nonlinear adsorption behavior</li>
    <li>Remove constant or highly correlated features (optional)</li>
    <li>Optionally apply PCA for dimensionality reduction</li>
  </ul>

  <p style="text-align: justify; font-family: cursive; font-size: 14px; color: green; font-weight: bold">
    3) Modeling
  </p>

  <ul>
    <li>Decision Tree</li>
    <li>SVR</li>
    <li>Random Forest</li>
    <li>Gradient Boosting</li>
    <li>XGBoost / LightGBM</li>
    <li>MLP (Neural Network)</li>
      <li>Voting and stacking</li>
      <li> Pipeline & ColumnTransformer</li>
  </ul>

  <p style="text-align: justify; font-family: cursive; font-size: 14px; color: green; font-weight: bold">
    4) Hyperparameter Tuning
  </p>

  <ul>
    <li>GridSearchCV or RandomizedSearchCV</li>
    <li>Key hyperparameters: n_estimators, max_depth, learning_rate, C, gamma, hidden_layer_sizes, alpha</li>
  </ul>

  <p style="text-align: justify; font-family: cursive; font-size: 14px; color: green; font-weight: bold">

    ✅ Outcome / Conclusion
  </p>

  <p>

Machine Learning Methodology
</h3>

<p style="text-align: justify;">
A supervised machine learning framework was developed to predict CO₂ uptake (wt%) of metal–organic frameworks (MOFs) using experimentally derived physicochemical descriptors. Prior to model development, the dataset was randomly divided into training (70%) and testing (30%) subsets to ensure unbiased performance evaluation and generalization assessment.
</p>

<p style="text-align: justify;">
Numerical features were preprocessed using mean imputation followed by standardization, while categorical features were imputed using the most frequent category and encoded via one-hot encoding. All preprocessing operations were integrated within a unified pipeline architecture to prevent data leakage during cross-validation and hyperparameter optimization.
</p>

<p style="text-align: justify;">
Multiple regression algorithms were evaluated, including Random Forest (RF), Support Vector Regression (SVR), and Extreme Gradient Boosting (XGBoost). Model hyperparameters were optimized using a combination of grid search and Bayesian optimization under five-fold cross-validation, with root mean squared error (RMSE) employed as the primary optimization metric.
</p>

<p style="text-align: justify;">
To further enhance predictive accuracy, a stacking ensemble model was constructed by integrating RF, XGBoost, and SVR as base learners, with ridge regression employed as the meta-learner. Model performance was quantitatively assessed using RMSE, mean absolute error (MAE), and the coefficient of determination (R²).
</p>

<p style="text-align: justify;">
Model interpretability was achieved using SHapley Additive exPlanations (SHAP) for tree-based models, enabling quantitative analysis of feature contributions to CO₂ uptake predictions. For non-tree-based models, permutation feature importance was applied to evaluate the relative influence of individual descriptors.
</p>

<p style="text-align: justify;">
This integrated machine learning framework ensures robust generalization, interpretability, and reproducibility, providing a reliable data-driven approach for the screening and rational design of MOF materials for carbon capture applications.
</p>

```python
def clean_and_format_dataset(df):
    # Step 1: Replace commas with dots in ALL cells
    df = df.applymap(lambda x: str(x).replace(',', '.') if isinstance(x, str) else x)

    # Step 2: Try to convert each column to numeric
    # for col in df.columns:
        # df[col] = pd.to_numeric(df[col], errors='ignore')

    # Step 3: Round numeric columns to 2 decimals
    for col in df.select_dtypes(include=['float', 'int']).columns:
        df[col] = df[col].round(2)

    # Step 4: Format numeric columns with commas and 2 decimals
    # for col in df.select_dtypes(include=['float', 'int']).columns:
       #  df[col] = df[col].apply(lambda x: "{:,.2f}".format(x))

    return df
```

XGBoost Overview:-
</p>

<p>
<b>XGBoost</b>, or <b>eXtreme Gradient Boosting</b>, is a powerful machine learning algorithm that belongs to the
family of gradient boosting methods and can be used for both <b>classification</b> and <b>regression</b> tasks.
It works by building a series of decision trees sequentially, where each new tree corrects the errors of
the previous ones, thereby improving predictive accuracy.
</p>


<p style="text-align: justify; font-family: cursive; font-size: 14px; color: green; font-weight: bold;">
Key Components of XGBoost:-
</p>

<ul>
  <li>
    <b>Boosting Technique:</b>
    XGBoost is an ensemble learning method that combines predictions from multiple weak learners
    (typically decision trees) to form a strong predictive model.
  </li>

  <li>
    <b>Objective Function:</b>
    XGBoost uses a regularized objective function composed of a loss function, which measures
    prediction error, and a regularization term that helps prevent overfitting.
  </li>

  <li>
    <b>Gradient Boosting:</b>
    The algorithm starts with a simple model and iteratively adds decision trees. Each new tree
    is trained on the residuals (errors) of the previous ensemble, minimizing the overall objective
    function through gradient-based optimization.
  </li>

  <li>
    <b>Tree Pruning:</b>
    XGBoost employs tree pruning strategies to limit tree depth, controlling model complexity
    and reducing the risk of overfitting.
  </li>

  <li>
    <b>Regularization:</b>
    Both <b>L1 (Lasso)</b> and <b>L2 (Ridge)</b> regularization are incorporated to penalize overly complex
    models. This improves generalization performance and supports effective feature selection.
  </li>

  <li>
    <b>Parallel Processing and Optimization:</b>
    XGBoost is optimized for speed and scalability, supporting parallel computation 
  </li>

  <li>
    <b>Feature Importance:</b>
    XGBoost provides feature importance scores that quantify the contribution of each feature
    to the model’s predictions, enabling better interpretability and insight into influential variables.
  </li>
</ul>

```python
def ZF_RF_Regression(X_train, X_test, y_train, y_test):
    """
    Random Forest Regression Model
    Includes full evaluation and visualization similar to ANN function.
    """

    # ----------------------------
    # 1️⃣ Build and train Random Forest model
    model_rf = RandomForestRegressor(
    max_depth=8,
    max_features='sqrt',
    min_samples_leaf=1,
    min_samples_split=2,
    n_estimators=300,
    random_state=42
    )
    model_rf.fit(X_train, y_train)

    # ----------------------------
    # 2️⃣ Predictions
    y_train_pred = model_rf.predict(X_train)
    y_test_pred = model_rf.predict(X_test)

    # ----------------------------
    # 3️⃣ Metrics
    metrics_train_rf = {"RMSE": np.sqrt(mean_squared_error(y_train, y_train_pred)),
                     "MAE": mean_absolute_error(y_train, y_train_pred),
                     "R2": r2_score(y_train, y_train_pred)}
    metrics_test_rf = {"RMSE": np.sqrt(mean_squared_error(y_test, y_test_pred)),
                    "MAE": mean_absolute_error(y_test, y_test_pred),
                    "R2": r2_score(y_test, y_test_pred)}

    print("=== RF Train Metrics ===", metrics_train_rf)
    print("=== RF Test Metrics ===", metrics_test_rf)


    Error_rf = pd.Series(metrics_test_rf,
    index=["RMSE", "MAE", "R2"],
    name="Random Forest")
    print(Error_rf)

    #5-Fold Cross-Validation (RMSE)
    cv_rmse_scores = cross_validate(
    model_rf,
    X_train,
    y_train,
    cv=5,
    scoring='neg_root_mean_squared_error',
    n_jobs=-1)

    cv_rmse2 = -cv_rmse_scores['test_score'].mean()
    print("\nRandom Forest Regression Results:")
    print("Cross validation-RSME:",cv_rmse2)


    sq_diff = np.mean((y_train_pred.mean() - y_test_pred.mean())**2)
    print(f"Train/Test Mean Squared Difference: {sq_diff:.4f}")

    # ----------------------------
    # Residuals
    residuals = y_test - y_test_pred
    standardized_residuals = residuals / np.std(residuals)

    # Residual plot
    plt.figure(figsize=(6, 5))
    plt.scatter(y_test_pred, residuals, alpha=0.6)
    plt.axhline(0, color="red", linestyle="--")
    plt.xlabel("Predicted")
    plt.ylabel("Residuals")
    plt.title("Residual Plot (RF)")
    plt.show()

    # Standardized Residuals
    plt.figure(figsize=(8, 6))
    sns.scatterplot(x=y_test_pred, y=standardized_residuals, color='purple', alpha=0.6)
    plt.axhline(0, color='black', linestyle='--')
    plt.axhline(2, color='red', linestyle='--')
    plt.axhline(-2, color='red', linestyle='--')
    plt.title("Standardized Residuals (RF)")
    plt.show()

    # ----------------------------
    # Predicted vs Actual (linear)
    plt.figure(figsize=(8, 6))
    max_val = max(y_test.max(), y_test_pred.max())
    plt.scatter(y_test, y_test_pred, color='red', alpha=0.6, label='Predicted')
    plt.plot([0, max_val], [0, max_val], color='green', linestyle='--', label='Perfect Prediction')
    plt.xlabel("Actual Hydrogen Flow Rate")
    plt.ylabel("Predicted Hydrogen Flow Rate")
    plt.title("RF: Predicted vs Actual")
    plt.legend()
    plt.show()

    # ----------------------------
    # Q-Q plot
    plt.figure(figsize=(8, 6))
    stats.probplot(residuals, dist="norm", plot=plt)
    plt.title('Q-Q Plot of Residuals (RF)')
    plt.show()

    # ----------------------------
    # Residual outliers
    threshold = 2 * np.std(residuals)
    outliers = residuals[np.abs(residuals) > threshold]
    plt.figure(figsize=(8, 6))
    sns.scatterplot(x=y_test_pred, y=residuals, color='blue', alpha=0.6, label='Residuals')
    plt.scatter(y_test_pred[np.abs(residuals) > threshold], outliers, color='red', label='Outliers')
    plt.axhline(0, color='black', linestyle='--')
    plt.title("Residuals Highlighting Outliers (RF)")
    plt.xlabel("Predicted Values")
    plt.ylabel("Residuals")
    plt.legend()
    plt.show()
    
    # Flatten
    y_test_flat = y_test
    pred_dt_reg_flat = y_test_pred

    # Create scatter plot
    plt.figure(figsize=(8,6))
    sns.scatterplot(x=range(len(y_test_flat)), y=y_test_flat, color='green', label='Actual')
    sns.scatterplot(x=range(len(pred_dt_reg_flat)), y=pred_dt_reg_flat, color='blue', label='Predicted')
    plt.xlabel('Sample Index')
    plt.ylabel('Value')
    plt.title('Actual vs Predicted Values')
    plt.legend()
    plt.grid(True)
    plt.show()

    return model_rf, metrics_train_rf, metrics_train_rf, sq_diff
```

<h3 style="color:green; font-weight:bold;">Bagging (Bootstrap Aggregating) & Boosting</h3>

<table style="width:100%; border-collapse:collapse; text-align:center;">
<tr style="background:#DFF0D8; font-weight:bold;">


  <p><strong>Bagging</strong> is an ensemble learning technique that combines multiple models to improve prediction accuracy and reduce overfitting. In this example, we used <strong>Random Forest</strong>, a popular implementation of bagging.</p>

  <h3>Key Characteristics of Bagging:</h3>
  <ol>
    <li>Creates multiple subsets of the original dataset using random sampling with replacement (bootstrap sampling).</li>
    <li>Trains a separate model on each subset.</li>
    <li>Combines predictions from all models through voting (for classification) or averaging (for regression).</li>
    <li>Reduces variance and helps to avoid overfitting.</li>
  </ol>

  <p><strong>Boosting</strong> is another ensemble learning technique that combines multiple weak learners to create a strong learner. In this example, we used <strong>Gradient Boosting</strong>.</p>

  <h3>Key Characteristics of Boosting:</h3>
  <ol>
    <li>Trains models sequentially, with each new model focusing on the errors of the previous models.</li>
    <li>Assigns higher weights to misclassified samples in subsequent iterations.</li>
    <li>Combines predictions from all models through weighted voting or averaging.</li>
    <li>Reduces both bias and variance, often achieving higher accuracy than bagging.</li>
  </ol>

  <h3>Main Differences between Bagging and Boosting:</h3>
  <ul>
    <li><strong>Model Building:</strong> Bagging builds multiple independent models in parallel; Boosting builds models sequentially, with each new model correcting errors of the previous ones.</li>
    <li><strong>Sample Weighting:</strong> Bagging treats all samples equally; Boosting assigns higher weights to misclassified samples in subsequent iterations.</li>
    <li><strong>Error Focus:</strong> Bagging trains each model independently focusing on the entire dataset; Boosting focuses on errors of the previous models.</li>
    <li><strong>Bias-Variance Tradeoff:</strong> Bagging primarily reduces variance; Boosting reduces both bias and variance.</li>
    <li><strong>Sensitivity to Noisy Data & Outliers:</strong> Bagging is more robust; Boosting can be sensitive.</li>
    <li><strong>Computational Complexity:</strong> Bagging can be parallelized and faster for large datasets; Boosting is sequential and harder to parallelize.</li>
  </ul>

  <p>In the provided code, <strong>Random Forest</strong> was used as an example of bagging and <strong>Gradient Boosting</strong> as an example of boosting. Both algorithms are applied to the wine dataset, and their performance is evaluated using accuracy scores and classification reports.</p>

```python


def ZF_GradientBoost_Regression(X_train, X_test, y_train, y_test, cv=5):
    """
    Gradient Boosting Regression with full evaluation and visualization.
    """

    # ----------------------------
    # 1️⃣ Build Gradient Boosting model
    model_gb = GradientBoostingRegressor(
        n_estimators=500,
        learning_rate=0.05,
        max_depth=3,              # reduce depth to prevent overfitting
        min_samples_split=5,      # regularization
        min_samples_leaf=3,       # regularization
        subsample=0.8,            # bagging effect
        validation_fraction=0.2,  # for early stopping
        n_iter_no_change=20,      # early stopping
        random_state=42
    )

    # ----------------------------
    # 2️⃣ Train model
    model_gb.fit(X_train, y_train)

    # ----------------------------
    # 3️⃣ Predictions
    y_train_pred_gb = model_gb.predict(X_train)
    y_test_pred_gb = model_gb.predict(X_test)

    # ----------------------------
    # 4️⃣ Metrics
    metrics_train = {
        "RMSE": np.sqrt(mean_squared_error(y_train, y_train_pred_gb)),
        "MAE": mean_absolute_error(y_train, y_train_pred_gb),
        "R2": r2_score(y_train, y_train_pred_gb)
    }
    metrics_test = {
        "RMSE": np.sqrt(mean_squared_error(y_test, y_test_pred_gb)),
        "MAE": mean_absolute_error(y_test, y_test_pred_gb),
        "R2": r2_score(y_test, y_test_pred_gb)
    }

    print("=== Gradient Boost Train Metrics ===", metrics_train)
    print("=== Gradient Boost Test Metrics ===", metrics_test)

    # ----------------------------
    # 5️⃣ Cross-validation
    scoring = {'MSE': 'neg_mean_squared_error', 'R2': 'r2'}
    cv_results = cross_validate(model_gb, X_train, y_train, cv=cv, scoring=scoring)
    print(f"{cv}-Fold CV Mean MSE: {-cv_results['test_MSE'].mean():.4f}")
    print(f"{cv}-Fold CV Mean R²: {cv_results['test_R2'].mean():.4f}")

    # ----------------------------
    # 6️⃣ Train/Test Mean Squared Difference
    sq_diff = np.mean((y_train_pred_gb.mean() - y_test_pred_gb.mean())**2)
    print(f"Train/Test Mean Squared Difference: {sq_diff:.4f}")

    # ----------------------------
    # 7️⃣ Residuals
    residuals = y_test - y_test_pred_gb
    standardized_residuals = residuals / np.std(residuals)

    plt.figure(figsize=(6, 5))
    plt.scatter(y_test_pred_gb, residuals, alpha=0.6)
    plt.axhline(0, color="red", linestyle="--")
    plt.xlabel("Predicted")
    plt.ylabel("Residuals")
    plt.title("Residual Plot (Gradient Boost)")
    plt.show()

    plt.figure(figsize=(8, 6))
    sns.scatterplot(x=y_test_pred_gb, y=standardized_residuals, color='purple', alpha=0.6)
    plt.axhline(0, color='black', linestyle='--')
    plt.axhline(2, color='red', linestyle='--')
    plt.axhline(-2, color='red', linestyle='--')
    plt.title("Standardized Residuals (Gradient Boost)")
    plt.show()

    # ----------------------------
    # 8️⃣ Predicted vs Actual (linear)
    plt.figure(figsize=(8, 6))
    max_val = max(y_test.max(), y_test_pred_gb.max())
    plt.scatter(y_test, y_test_pred_gb, color='red', alpha=0.6, label='Predicted')
    plt.plot([0, max_val], [0, max_val], color='green', linestyle='--', label='Perfect Prediction')
    plt.xlabel("Actual")
    plt.ylabel("Predicted")
    plt.title("Gradient Boost: Predicted vs Actual")
    plt.legend()
    plt.show()

    # ----------------------------
    # 9️⃣ Q-Q plot
    plt.figure(figsize=(8, 6))
    stats.probplot(residuals, dist="norm", plot=plt)
    plt.title('Q-Q Plot of Residuals (Gradient Boost)')
    plt.show()

    # ----------------------------
    # 🔟 Residual outliers
    threshold = 2 * np.std(residuals)
    outliers = residuals[np.abs(residuals) > threshold]
    plt.figure(figsize=(8, 6))
    sns.scatterplot(x=y_test_pred_gb, y=residuals, color='blue', alpha=0.6, label='Residuals')
    plt.scatter(y_test_pred_gb[np.abs(residuals) > threshold], outliers, color='red', label='Outliers')
    plt.axhline(0, color='black', linestyle='--')
    plt.title("Residuals Highlighting Outliers (Gradient Boost)")
    plt.xlabel("Predicted Values")
    plt.ylabel("Residuals")
    plt.legend()
    plt.show()

    # ----------------------------
    # 1️⃣1️⃣ Actual vs Predicted time-series style
    plt.figure(figsize=(8,6))
    sns.scatterplot(x=range(len(y_test)), y=y_test, color='green', label='Actual')
    sns.scatterplot(x=range(len(y_test_pred_gb)), y=y_test_pred_gb, color='blue', label='Predicted')
    plt.xlabel('Sample Index')
    plt.ylabel('Value')
    plt.title('Actual vs Predicted Values')
    plt.legend()
    plt.grid(True)
    plt.show()

    return model_gb, metrics_train, metrics_test, sq_diff
```

<h3 style="color:green; font-weight:bold;">SVR Hyperparameters Explained</h3>

<ol>
  <li>
    <b>kernel='rbf'</b>
    <p>RBF (Radial Basis Function) is the most common kernel.</p>
    <p>It allows the model to learn <b>non-linear relationships</b>.</p>
  </li>

  <li>
    <b>C = 100</b>
    <p><b>C</b> is the regularization parameter.</p>
    <p><b>High C (like 100)</b> → less regularization.</p>
    <p>The model tries to fit training data more strictly, which can cause <b>overfitting</b>.</p>
    <p><b>Low C</b> → smoother model, more generalization.</p>
  </li>

  <li>
    <b>gamma = 0.1</b>
    <p>Controls how far the influence of a single training example reaches.</p>
    <p><b>High gamma</b> → very tight fit around training points (overfitting risk).</p>
    <p><b>Low gamma</b> → smoother decision boundary.</p>
  </li>

  <li>
    <b>epsilon = 0.1</b>
    <p>Defines the <b>tube</b> around the regression line.</p>
    <p>Predictions within ±0.1 error are ignored (no penalty).</p>
    <p><b>Higher epsilon</b> → simpler model (less overfitting).</p>
  </li>
</ol>

</div>


```python


def ZF_SVR_Regression(X_train, X_test, y_train, y_test, cv=5):
    """
    SVR Regression with full evaluation and visualization.
    """

    # ----------------------------
    # 1️⃣ Build SVR model
    model_svr = SVR(
        kernel='rbf',
        C=100,
        gamma=0.1,
        epsilon=0.1
    )

    # ----------------------------
    # 2️⃣ Train model
    model_svr.fit(X_train, y_train)

    # ----------------------------
    # 3️⃣ Predictions
    y_train_pred = model_svr.predict(X_train)
    y_test_pred = model_svr.predict(X_test)

    # ----------------------------
    # 4️⃣ Metrics
    metrics_train = {
        "RMSE": np.sqrt(mean_squared_error(y_train, y_train_pred)),
        "MAE": mean_absolute_error(y_train, y_train_pred),
        "R2": r2_score(y_train, y_train_pred)
    }
    metrics_test = {
        "RMSE": np.sqrt(mean_squared_error(y_test, y_test_pred)),
        "MAE": mean_absolute_error(y_test, y_test_pred),
        "R2": r2_score(y_test, y_test_pred)
    }

    print("=== SVR Train Metrics ===", metrics_train)
    print("=== SVR Test Metrics ===", metrics_test)

    # ----------------------------
    # 5️⃣ Cross-validation
    scoring = {'MSE': 'neg_mean_squared_error', 'R2': 'r2'}
    cv_results = cross_validate(model_svr, X_train, y_train, cv=cv, scoring=scoring)
    print(f"{cv}-Fold CV Mean MSE: {-cv_results['test_MSE'].mean():.4f}")
    print(f"{cv}-Fold CV Mean R²: {cv_results['test_R2'].mean():.4f}")

    # ----------------------------
    # 6️⃣ Train/Test Mean Squared Difference
    sq_diff = np.mean((y_train_pred.mean() - y_test_pred.mean())**2)
    print(f"Train/Test Mean Squared Difference: {sq_diff:.4f}")

    # ----------------------------
    # 7️⃣ Residuals
    residuals = y_test - y_test_pred
    standardized_residuals = residuals / np.std(residuals)

    plt.figure(figsize=(6, 5))
    plt.scatter(y_test_pred, residuals, alpha=0.6)
    plt.axhline(0, color="red", linestyle="--")
    plt.xlabel("Predicted")
    plt.ylabel("Residuals")
    plt.title("Residual Plot (SVR)")
    plt.show()

    plt.figure(figsize=(8, 6))
    sns.scatterplot(x=y_test_pred, y=standardized_residuals, alpha=0.6)
    plt.axhline(0, color='black', linestyle='--')
    plt.axhline(2, color='red', linestyle='--')
    plt.axhline(-2, color='red', linestyle='--')
    plt.title("Standardized Residuals (SVR)")
    plt.show()

    # ----------------------------
    # 8️⃣ Predicted vs Actual (linear)
    plt.figure(figsize=(8, 6))
    max_val = max(y_test.max(), y_test_pred.max())
    plt.scatter(y_test, y_test_pred, color='red', alpha=0.6, label='Predicted')
    plt.plot([0, max_val], [0, max_val], color='green', linestyle='--', label='Perfect Prediction')
    plt.xlabel("Actual")
    plt.ylabel("Predicted")
    plt.title("SVR: Predicted vs Actual")
    plt.legend()
    plt.show()

    # ----------------------------
    # 9️⃣ Q-Q plot
    plt.figure(figsize=(8, 6))
    stats.probplot(residuals, dist="norm", plot=plt)
    plt.title('Q-Q Plot of Residuals (SVR)')
    plt.show()

    # ----------------------------
    # 🔟 Residual outliers
    threshold = 2 * np.std(residuals)
    outliers = residuals[np.abs(residuals) > threshold]
    plt.figure(figsize=(8, 6))
    sns.scatterplot(x=y_test_pred, y=residuals, alpha=0.6)
    plt.scatter(y_test_pred[np.abs(residuals) > threshold], outliers, color='red')
    plt.axhline(0, color='black', linestyle='--')
    plt.title("Residuals Highlighting Outliers (SVR)")
    plt.xlabel("Predicted Values")
    plt.ylabel("Residuals")
    plt.show()

    # ----------------------------
    # 1️⃣1️⃣ Actual vs Predicted time-series style
    plt.figure(figsize=(8,6))
    sns.scatterplot(x=range(len(y_test)), y=y_test, color='green', label='Actual')
    sns.scatterplot(x=range(len(y_test_pred)), y=y_test_pred, color='blue', label='Predicted')
    plt.xlabel('Sample Index')
    plt.ylabel('Value')
    plt.title('Actual vs Predicted Values')
    plt.legend()
    plt.grid(True)
    plt.show()

    return model_svr, metrics_train, metrics_test, sq_diff
```

```python

def ZF_MLP_Regression(X_train, X_test, y_train, y_test, cv=5):
    """
    MLP Regression with full evaluation and visualization.
    """

    # ----------------------------
    # 1️⃣ Build MLP model
    model_mlp = MLPRegressor(
        hidden_layer_sizes=(100, 50),
        activation='relu',
        solver='adam',
        learning_rate='adaptive',
        learning_rate_init=0.001,
        max_iter=1000,
        random_state=42
    )

    # ----------------------------
    # 2️⃣ Train model
    model_mlp.fit(X_train, y_train)

    # ----------------------------
    # 3️⃣ Predictions
    y_train_pred = model_mlp.predict(X_train)
    y_test_pred = model_mlp.predict(X_test)

    # ----------------------------
    # 4️⃣ Metrics
    metrics_train = {
        "RMSE": np.sqrt(mean_squared_error(y_train, y_train_pred)),
        "MAE": mean_absolute_error(y_train, y_train_pred),
        "R2": r2_score(y_train, y_train_pred)
    }
    metrics_test = {
        "RMSE": np.sqrt(mean_squared_error(y_test, y_test_pred)),
        "MAE": mean_absolute_error(y_test, y_test_pred),
        "R2": r2_score(y_test, y_test_pred)
    }

    print("=== MLP Train Metrics ===", metrics_train)
    print("=== MLP Test Metrics ===", metrics_test)

    # ----------------------------
    # 5️⃣ Cross-validation
    scoring = {'MSE': 'neg_mean_squared_error', 'R2': 'r2'}
    cv_results = cross_validate(model_mlp, X_train, y_train, cv=cv, scoring=scoring)
    print(f"{cv}-Fold CV Mean MSE: {-cv_results['test_MSE'].mean():.4f}")
    print(f"{cv}-Fold CV Mean R²: {cv_results['test_R2'].mean():.4f}")

    # ----------------------------
    # 6️⃣ Train/Test Mean Squared Difference
    sq_diff = np.mean((y_train_pred.mean() - y_test_pred.mean())**2)
    print(f"Train/Test Mean Squared Difference: {sq_diff:.4f}")

    # ----------------------------
    # 7️⃣ Residuals
    residuals = y_test - y_test_pred
    standardized_residuals = residuals / np.std(residuals)

    plt.figure(figsize=(6, 5))
    plt.scatter(y_test_pred, residuals, alpha=0.6)
    plt.axhline(0, color="red", linestyle="--")
    plt.xlabel("Predicted")
    plt.ylabel("Residuals")
    plt.title("Residual Plot (MLP)")
    plt.show()

    plt.figure(figsize=(8, 6))
    sns.scatterplot(x=y_test_pred, y=standardized_residuals, alpha=0.6)
    plt.axhline(0, color='black', linestyle="--")
    plt.axhline(2, color='red', linestyle="--")
    plt.axhline(-2, color='red', linestyle="--")
    plt.title("Standardized Residuals (MLP)")
    plt.show()

    # ----------------------------
    # 8️⃣ Predicted vs Actual (linear)
    plt.figure(figsize=(8, 6))
    max_val = max(y_test.max(), y_test_pred.max())
    plt.scatter(y_test, y_test_pred, color='red', alpha=0.6, label='Predicted')
    plt.plot([0, max_val], [0, max_val], color='green', linestyle="--", label='Perfect Prediction')
    plt.xlabel("Actual")
    plt.ylabel("Predicted")
    plt.title("MLP: Predicted vs Actual")
    plt.legend()
    plt.show()

    # ----------------------------
    # 9️⃣ Q-Q plot
    plt.figure(figsize=(8, 6))
    stats.probplot(residuals, dist="norm", plot=plt)
    plt.title('Q-Q Plot of Residuals (MLP)')
    plt.show()

    # ----------------------------
    # 🔟 Residual outliers
    threshold = 2 * np.std(residuals)
    outliers = residuals[np.abs(residuals) > threshold]
    plt.figure(figsize=(8, 6))
    sns.scatterplot(x=y_test_pred, y=residuals, alpha=0.6)
    plt.scatter(y_test_pred[np.abs(residuals) > threshold], outliers, color='red')
    plt.axhline(0, color='black', linestyle="--")
    plt.title("Residuals Highlighting Outliers (MLP)")
    plt.xlabel("Predicted Values")
    plt.ylabel("Residuals")
    plt.show()

    # ----------------------------
    # 1️⃣1️⃣ Actual vs Predicted time-series style
    plt.figure(figsize=(8,6))
    sns.scatterplot(x=range(len(y_test)), y=y_test, color='green', label='Actual')
    sns.scatterplot(x=range(len(y_test_pred)), y=y_test_pred, color='blue', label='Predicted')
    plt.xlabel('Sample Index')
    plt.ylabel('Value')
    plt.title('Actual vs Predicted Values')
    plt.legend()
    plt.grid(True)
    plt.show()

    return model_mlp, metrics_train, metrics_test, sq_diff
```

```python
def ZF_ANN_Regression(X_train, X_test, y_train, y_test, epochs=150, batch_size=32, feature_names=None):

    #  Build ANN model
    ann_model = Sequential([
        Dense(32, input_dim=X_train.shape[1], activation='relu', kernel_regularizer=regularizers.l2(0.01)),
        Dense(16, activation='relu'),
        Dense(1, activation='linear')
    ])
    ann_model.compile(loss='mean_squared_error', optimizer=Adam(), metrics=['mse', 'mae'])
    ann_model.summary()

    # Early stopping
    early_stop = callbacks.EarlyStopping(monitor='val_loss', patience=15, restore_best_weights=True)

    # Fit model
    history = ann_model.fit(
        X_train, y_train.ravel(),
        validation_data=(X_test, y_test.ravel()),
        epochs=epochs,
        batch_size=batch_size,
        verbose=1,
        callbacks=[early_stop]
    )

    # Predictions
    y_train_pred_mlp = ann_model.predict(X_train).ravel()
    y_test_pred_mlp = ann_model.predict(X_test).ravel()

    #  Metrics
    metrics_train = {
        "RMSE": np.sqrt(mean_squared_error(y_train.ravel(), y_train_pred_mlp)),
        "MAE": mean_absolute_error(y_train.ravel(), y_train_pred_mlp),
        "R2": r2_score(y_train.ravel(), y_train_pred_mlp)
    }
    metrics_test = {
        "RMSE": np.sqrt(mean_squared_error(y_test.ravel(), y_test_pred_mlp)),
        "MAE": mean_absolute_error(y_test.ravel(), y_test_pred_mlp),
        "R2": r2_score(y_test.ravel(), y_test_pred_mlp)
    }
    print("=== ANN Train Metrics ===", metrics_train)
    print("=== ANN Test Metrics ===", metrics_test)

    # Train/Test mean difference
    sq_diff = np.mean((y_train_pred_mlp.mean() - y_test_pred_mlp.mean())**2)
    print(f"Train/Test Mean Squared Difference: {sq_diff:.4f}")

    #  Residual Plot
    plt.figure(figsize=(6,5))
    plt.scatter(y_test_pred_mlp, y_test.ravel() - y_test_pred_mlp, alpha=0.6)
    plt.axhline(0, color="red", linestyle="--")
    plt.xlabel("Predicted")
    plt.ylabel("Residuals")
    plt.title("Residual Plot (ANN)")
    plt.show()

    # 
    #  Training History Plots
    epochs_range = range(1, len(history.history['loss']) + 1)
    plt.figure(figsize=(8,5))
    plt.plot(epochs_range, history.history['loss'], 'b-', label='Training Loss')
    plt.plot(epochs_range, history.history['val_loss'], 'r-', label='Validation Loss')
    plt.xlabel('Epochs')
    plt.ylabel('Loss (MSE)')
    plt.title('Training and Validation Loss')
    plt.legend()
    plt.show()

    if 'mae' in history.history:
        plt.figure(figsize=(8,5))
        plt.plot(epochs_range, history.history['mae'], 'g-', label='Training MAE')
        plt.plot(epochs_range, history.history['val_mae'], 'k-', label='Validation MAE')
        plt.xlabel('Epochs')
        plt.ylabel('Mean Absolute Error')
        plt.title('Training and Validation MAE')
        plt.legend()
        plt.show()

    # ️⃣ Predicted vs Actual
    plt.figure(figsize=(8,6))
    x_ax = range(len(y_test_pred_mlp))
    plt.scatter(x_ax, y_test.ravel(), s=5, color="blue", label="Actual")
    plt.plot(x_ax, y_test_pred_mlp, lw=0.8, color="red", label="Predicted")
    plt.legend()
    plt.xlabel("Sample Index")
    plt.ylabel("CO$_2$ Adsorbed (mmol/g)")
    plt.title("ANN: Actual vs Predicted CO$_2$ Adsorption")
    plt.show()

    
    #  Joint Plot
    g = sns.JointGrid(x=y_test.ravel(), y=y_test_pred_mlp, height=8)
    sns.scatterplot(x=y_train.ravel(), y=y_train_pred_mlp, s=100, color='orange', ax=g.ax_joint, label='Train')
    sns.scatterplot(x=y_test.ravel(), y=y_test_pred_mlp, s=100, color='blue', ax=g.ax_joint, label='Test')
    sns.regplot(x=y_test.ravel(), y=y_test_pred_mlp, ax=g.ax_joint, scatter=False, color='blue')
    g.set_axis_labels("Actual CO$_2$ adsorbed (mmol/g)", "Predicted CO$_2$ adsorbed (mmol/g)", fontsize=16)
    sns.histplot(y_train.ravel(), ax=g.ax_marg_x, color='orange')
    sns.histplot(y_test.ravel(), ax=g.ax_marg_x, color='blue')
    g.ax_marg_x.legend(["Train", "Test"])
    plt.suptitle("ANN: Actual vs Predicted CO$_2$ Adsorption", fontsize=18, y=1.02)
    plt.show()

    
    #  Error Distribution
    errors = y_test.ravel() - y_test_pred_mlp
    plt.figure(figsize=(8, 5))
    plt.hist(errors, bins=30, alpha=0.7, edgecolor='black')
    plt.axvline(0, color='red', linestyle='--')
    plt.xlabel('Prediction Error')
    plt.ylabel('Frequency')
    plt.title('Distribution of Prediction Errors (ANN)')
    plt.show()

    #  Explicit Evaluation Metrics Bar Chart
    mae_nn = metrics_test['MAE']
    mse_nn = mean_squared_error(y_test.ravel(), y_test_pred_mlp)
    rmse_nn = metrics_test['RMSE']
    r2_nn = metrics_test['R2']

    plt.figure(figsize=(6, 4))
    values = [mae_nn, mse_nn, r2_nn, rmse_nn]
    bars = plt.bar(['MAE', 'MSE', 'R²', 'RMSE'], values, color='brown')
    for bar, val in zip(bars, values):
        plt.text(bar.get_x() + bar.get_width()/2, bar.get_height()*1.01, f'{val:.4f}',
                 ha='center', va='bottom', fontsize=11)
    plt.title('Neural Network - Evaluation Metrics')
    plt.show()

    return ann_model, metrics_train, metrics_test, sq_diff, history
```

```python
# 1) define models
models = {
    "Decision Tree": DecisionTreeRegressor(max_depth=5, random_state=42, min_samples_split=8, min_samples_leaf=5, splitter="best"),
    "Random Forest": RandomForestRegressor(max_depth=8, max_features='sqrt',min_samples_leaf=1, min_samples_split=2,n_estimators=300, random_state=42),
    "GBoost": GradientBoostingRegressor(n_estimators=500,learning_rate=0.05, max_depth=3,min_samples_split=5, min_samples_leaf=3,subsample=0.8,            
        validation_fraction=0.2,n_iter_no_change=20,random_state=42),
    "XGBoost": XGBRegressor(n_estimators=100, learning_rate=0.1, random_state=42),
    "SVR": SVR( kernel='rbf',C=100,gamma=0.1, epsilon=0.1),
    "MLP": MLPRegressor(hidden_layer_sizes=(100, 50), activation='relu', solver='adam', learning_rate='adaptive',
                        learning_rate_init=0.001, max_iter=1000, random_state=42)
}

# 2) prepare storage
ks = range(2, 11)
results = {name: {'mse': [], 'rmse': [], 'r2': []} for name in models}

# 3) loop over k
for k in ks:
    cv = KFold(n_splits=k, shuffle=True, random_state=42)
    for name, model in models.items():
        cvres = cross_validate(
            model, X_train, y_train, cv=cv,
            scoring={
                'r2': 'r2',
                'mse': make_scorer(mean_squared_error)
            },
            return_train_score=False
        )
        mse_mean = np.mean(cvres['test_mse'])
        results[name]['mse'].append(mse_mean)
        results[name]['rmse'].append(np.sqrt(mse_mean))
        results[name]['r2'].append(np.mean(cvres['test_r2']))

# 4) plot each metric
for metric in ['mse', 'rmse', 'r2']:
    plt.figure(figsize=(8, 4))
    for name in models:
        plt.plot(ks, results[name][metric], marker='o', label=name)
    plt.title(f"{metric.upper()} vs. Number of Folds")
    plt.xlabel("k (folds)")
    plt.ylabel(metric.upper())
    plt.xticks(ks)
    plt.grid(True)
    plt.legend()
    plt.tight_layout()
    plt.show()
```

```python

def best_par_svr(X_train, y_train, X_test, y_test):
    results = {}

    # Imputer (fill NaN)
    imputer = SimpleImputer(strategy='mean')
    X_train_imp = imputer.fit_transform(X_train)
    X_test_imp = imputer.transform(X_test)

    for C in [0.1, 1, 10, 100]:
        for epsilon in [0.01, 0.1, 0.2]:
            for gamma in ['scale', 'auto', 0.001, 0.01, 0.1]:
                for kernel in ['linear', 'rbf']:

                    svr = SVR(
                        C=C,
                        epsilon=epsilon,
                        gamma=gamma,
                        kernel=kernel
                    )

                    # Fit model
                    svr.fit(X_train_imp, y_train)

                    # Evaluate
                    mse = mean_squared_error(y_test, svr.predict(X_test_imp))
                    rmse = mse ** 0.5
                    r2_train = svr.score(X_train_imp, y_train)
                    r2_test = svr.score(X_test_imp, y_test)

                    # Print results
                    print(f"C={C}, epsilon={epsilon}, gamma={gamma}, kernel={kernel}")
                    print(f"MSE on test set: {mse:.4f}")
                    print(f"RMSE on test set: {rmse:.4f}")
                    print(f"R² on training set: {r2_train:.3f}")
                    print(f"R² on test set: {r2_test:.3f}")
                    print("==============================================")

                    # Store results
                    results[f'{C}-{epsilon}-{gamma}-{kernel}'] = r2_test

    # Convert results to DataFrame
    df_scores = pd.DataFrame(results, index=[0]).T
    df_scores.columns = ['R2_test']
    return results, df_scores
```

```python

def evaluate_model(model, X_train, X_test, y_train, y_test):
    imputer = SimpleImputer(strategy='mean')
    X_train_imp = imputer.fit_transform(X_train)
    X_test_imp = imputer.transform(X_test)

    model.fit(X_train_imp, y_train)
    y_pred = model.predict(X_test_imp)

    rmse = np.sqrt(mean_squared_error(y_test, y_pred))
    mae = mean_absolute_error(y_test, y_pred)
    r2 = r2_score(y_test, y_pred)

    return rmse, mae, r2


# -------------------------
# MLP Hyperparameter Tuning
# -------------------------
hidden_layer_grid = [(50,), (100,), (100, 50), (100, 100), (150, 100, 50)]
activation_grid = ['relu', 'tanh', 'logistic']
solver_grid = ['adam', 'lbfgs']
alpha_grid = [0.0001, 0.001, 0.01]
learning_rate_grid = ['constant', 'adaptive']

# 1) hidden_layer_sizes tuning
rmse_list_hl, mae_list_hl, r2_list_hl = [], [], []
for hl in hidden_layer_grid:
    mlp = MLPRegressor(hidden_layer_sizes=hl, max_iter=1000, random_state=42)
    rmse, mae, r2 = evaluate_model(mlp, X_train, X_test, y_train, y_test)

    rmse_list_hl.append(rmse)
    mae_list_hl.append(mae)
    r2_list_hl.append(r2)

plt.figure(figsize=(10,5))
plt.plot([str(x) for x in hidden_layer_grid], rmse_list_hl, marker='o', label='RMSE')
plt.plot([str(x) for x in hidden_layer_grid], mae_list_hl, marker='s', label='MAE')
plt.plot([str(x) for x in hidden_layer_grid], r2_list_hl, marker='^', label='R²')
plt.xlabel('hidden_layer_sizes')
plt.ylabel('Score')
plt.legend()
plt.title('MLP - hidden_layer_sizes tuning')
plt.xticks(rotation=45)
plt.show()


# 2) activation tuning
rmse_list_act, mae_list_act, r2_list_act = [], [], []
for act in activation_grid:
    mlp = MLPRegressor(hidden_layer_sizes=(100,), activation=act, max_iter=1000, random_state=42)
    rmse, mae, r2 = evaluate_model(mlp, X_train, X_test, y_train, y_test)

    rmse_list_act.append(rmse)
    mae_list_act.append(mae)
    r2_list_act.append(r2)

plt.figure(figsize=(8,5))
plt.plot(activation_grid, rmse_list_act, marker='o', label='RMSE')
plt.plot(activation_grid, mae_list_act, marker='s', label='MAE')
plt.plot(activation_grid, r2_list_act, marker='^', label='R²')
plt.xlabel('activation')
plt.ylabel('Score')
plt.legend()
plt.title('MLP - activation tuning')
plt.show()


# 3) solver tuning
rmse_list_solver, mae_list_solver, r2_list_solver = [], [], []
for sol in solver_grid:
    mlp = MLPRegressor(hidden_layer_sizes=(100,), solver=sol, max_iter=1000, random_state=42)
    rmse, mae, r2 = evaluate_model(mlp, X_train, X_test, y_train, y_test)

    rmse_list_solver.append(rmse)
    mae_list_solver.append(mae)
    r2_list_solver.append(r2)

plt.figure(figsize=(8,5))
plt.plot(solver_grid, rmse_list_solver, marker='o', label='RMSE')
plt.plot(solver_grid, mae_list_solver, marker='s', label='MAE')
plt.plot(solver_grid, r2_list_solver, marker='^', label='R²')
plt.xlabel('solver')
plt.ylabel('Score')
plt.legend()
plt.title('MLP - solver tuning')
plt.show()


# 4) alpha tuning
rmse_list_alpha, mae_list_alpha, r2_list_alpha = [], [], []
for a in alpha_grid:
    mlp = MLPRegressor(hidden_layer_sizes=(100,), alpha=a, max_iter=1000, random_state=42)
    rmse, mae, r2 = evaluate_model(mlp, X_train, X_test, y_train, y_test)

    rmse_list_alpha.append(rmse)
    mae_list_alpha.append(mae)
    r2_list_alpha.append(r2)

plt.figure(figsize=(8,5))
plt.plot(alpha_grid, rmse_list_alpha, marker='o', label='RMSE')
plt.plot(alpha_grid, mae_list_alpha, marker='s', label='MAE')
plt.plot(alpha_grid, r2_list_alpha, marker='^', label='R²')
plt.xscale('log')
plt.xlabel('alpha')
plt.ylabel('Score')
plt.legend()
plt.title('MLP - alpha tuning')
plt.show()


# 5) learning_rate tuning
rmse_list_lr, mae_list_lr, r2_list_lr = [], [], []
for lr in learning_rate_grid:
    mlp = MLPRegressor(hidden_layer_sizes=(100,), learning_rate=lr, max_iter=1000, random_state=42)
    rmse, mae, r2 = evaluate_model(mlp, X_train, X_test, y_train, y_test)

    rmse_list_lr.append(rmse)
    mae_list_lr.append(mae)
    r2_list_lr.append(r2)

plt.figure(figsize=(8,5))
plt.plot(learning_rate_grid, rmse_list_lr, marker='o', label='RMSE')
plt.plot(learning_rate_grid, mae_list_lr, marker='s', label='MAE')
plt.plot(learning_rate_grid, r2_list_lr, marker='^', label='R²')
plt.xlabel('learning_rate')
plt.ylabel('Score')
plt.legend()
plt.title('MLP - learning_rate tuning')
plt.show()
```
